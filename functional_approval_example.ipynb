# Example 2: Non-rectangular approval functions

```python
# Create a society
T = Society('S2')  # "S2" is an arbitrary name for the society.  tick is optional; used only when plotting below

# Specify approval sets for that society
T.add_approval_set('A1', lambda x: 2 if (x >=0 and x<=0.5) else 0 ) # specify (1) name for the set and (2) approval function
T.add_approval_set('A2', lambda x: 1.5 if (x >=1 and x<=5/3) else 0 )
T.add_approval_set('A3', lambda x: 1 if (x >=4/3 and x<=7/3) else 0 )
T.add_approval_set('A4', lambda x: 0.5 if (x >=5/3 and x<=11/3) else 0 )
T.add_approval_set('A5', lambda x: 0.5 if (x >=2 and x<=4) else 0 )

# example of a non-rectangular function
T.add_approval_set('A6', lambda x: 2*(x-2) if (x >=2 and x<=3) else 0 )


# can also define a function first, then use it to define an approval set:
def triangle( x ):
    if x < 3 or x > 5:
        y = 0
    elif x >= 3 and x <= 4:
        y = 0.5*(x-3)
    elif x > 4 and x <= 5:
        y = 0.5*(5-x)
        
    return y
        
T.add_approval_set('A8', triangle)
```

```python
T.print_society_info()
```

```python
T.visualize()
```

```python
T.visualize(c = 0.25)
```

# Example 3

```python
S2 = Society()

# Specify approval sets for that society
S2.add_approval_set('A1', lambda x: 2 if (x >=0 and x<=0.5) else 0 ) # specify (1) name for the set and (2) approval function
S2.add_approval_set('A2', lambda x: 1.5 if (x >=3/3+0.01 and x<=5/3+0.01) else 0 )
S2.add_approval_set('A3', lambda x: 1.5 if (x >=6/3 and x<=8/3) else 0 )
S2.add_approval_set('A4', lambda x: 1 if (x >=4/3 and x<=7/3) else 0 )
S2.add_approval_set('A5', lambda x: 0.5 if (x >=5/3 and x<=11/3) else 0 )
S2.add_approval_set('A6', lambda x: 0.5 if (x >=7/3-0.01 and x<=13/3-0.01) else 0 )

S2.print_society_info()
k = 2
m = 3
is_agreeable, bad_m = S2.check_agreeability(k, m)
print(f'Is ({k}, {m})-agreeable? {is_agreeable}')
if bad_m:
    print(f' - Collection(s) of {m} sets whose count function is not >= {k}: {bad_m}')
```

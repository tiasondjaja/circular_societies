# Example 1

```python
# Create a society
S = Society('S1', tick=1/3)  # "S1" is an arbitrary name for the society.  tick is optional; used only when plotting below

# Specify approval sets for that society
S.add_approval_set('A1', lambda x: 2 if (x >=0 and x<=0.5) else 0 ) # specify (1) name for the set and (2) approval function
S.add_approval_set('A2', lambda x: 1.5 if (x >=1 and x<=5/3) else 0 )
S.add_approval_set('A3', lambda x: 1 if (x >=4/3 and x<=7/3) else 0 )
S.add_approval_set('A4', lambda x: 0.5 if (x >=5/3 and x<=11/3) else 0 )
S.add_approval_set('A5', lambda x: 0.5 if (x >=2 and x<=4) else 0 )
```

```python
S.print_society_info()
```

```python
S.visualize(xmin=0, xmax = 5, ymax=3)
```

```python
S.plot_count_function(xmin=0, xmax = 5, ymax=4)
```

# Example 2: Non-rectangular approval functions

```python
# Create a society
T = Society('S2', tick=1/3)  # "S2" is an arbitrary name for the society.  tick is optional; used only when plotting below

# Specify approval sets for that society
T.add_approval_set('A1', lambda x: 2 if (x >=0 and x<=0.5) else 0 ) # specify (1) name for the set and (2) approval function
T.add_approval_set('A2', lambda x: 1.5 if (x >=1 and x<=5/3) else 0 )
T.add_approval_set('A3', lambda x: 1 if (x >=4/3 and x<=7/3) else 0 )
T.add_approval_set('A4', lambda x: 0.5 if (x >=5/3 and x<=11/3) else 0 )
T.add_approval_set('A5', lambda x: 0.5 if (x >=2 and x<=4) else 0 )

# example of a non-rectangular function
T.add_approval_set('A6', lambda x: 2*(x-2) if (x >=2 and x<=3) else 0 )


# can also define a function first, then use it to define an approval set:
def triangle( x ):
    if x < 3 or x > 5:
        y = 0
    elif x >= 3 and x <= 4:
        y = 0.5*(x-3)
    elif x > 4 and x <= 5:
        y = 0.5*(5-x)
        
    return y
        
T.add_approval_set('A8', triangle)
```

```python
T.print_society_info()
```

```python
T.visualize(xmin=0, xmax = 5, ymax=3)
```

```python
T.plot_count_function(xmin=0, xmax = 5, ymax=4)
```
